```python
import SPaCE as sp
import numpy as np
import matplotlib.pyplot as plt
```

```python
## CHIRP hahn echo
test=sp.spin(50000,method='Uniform',width=1) 

trj=sp.trajectory(1000,2,test)
p90=sp.pulse(np.pi/2,200,'wurst',-0.25,0.25,32)
p180=sp.pulse(np.pi,100,'wurst',-0.25,0.25,32)

trj.flip(p90,0,'+x') #pi/2 pulse in the x-direction at time 0
trj.precess(200,400) #precession from t=30 to t=330
trj.flip(p180,400,'+x') #pi pulse in the x-direction at time 330 (300 ns after the first pulse)
trj.precess(500,1000) #precession from t=390 to t=1000
trj.calc_intensity() #Calc intensity based on transition. Only matters with spins of S>1/2. But still needs to be executed for all systems
trj.net_M() #Calculate the net magnetization
```
